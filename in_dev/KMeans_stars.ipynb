# Part 1: Plotting

The code below makes a plot of absolute magnitude vs. temperature for 100,000+ stars.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv("https://github.com/adamlamee/CODINGinK12/raw/master/data/stars.csv")
data.shape
```

```python
data.head(3)
```

```python
plt.scatter(data['temp'], data['absmag'], alpha=.2, s=.1)
plt.xlabel('Temperature (K)')
plt.ylabel('Absolute Magnitude')
plt.xlim(0, 15000)
plt.ylim(20,-15)
plt.show()
```

# Part 2: Clustering

The code below uses the k-means algorithm to identify clusters.

```python
# makes a data set with only temp and absmag, dropping NaN values that will cause problems with clustering
sample = data[['temp', 'absmag']].copy()
sample.dropna(inplace=True)
sample.shape
```

```python
# Becuase the two axes have such different ranges, clustering algorithms will
# often lose sight of variations in the smaller scaled quantity

# this recales the data so the clusters will weigh x and y equally
from sklearn.preprocessing import StandardScaler
scaled_data = StandardScaler().set_output().fit_transform(sample)
```

```python
kmeans = KMeans(n_clusters=4)

predictions = kmeans.fit_predict(scaled_data)

plt.scatter(sample['temp'], sample['absmag'], alpha=.2, s=.1, c=predictions)
plt.xlabel('Temperature (K)')
plt.ylabel('Absolute Magnitude')
plt.xlim(0, 15000)
plt.ylim(20,-15)
plt.show()
```
