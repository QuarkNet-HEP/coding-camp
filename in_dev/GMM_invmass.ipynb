# Part 1: Plotting

The code below makes a  histogram of invariant mass for 14,000+ dielectron events with background subtracted.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.mixture import GaussianMixture
from scipy import stats
from scipy.stats import norm
```

```python
data = pd.read_csv('https://github.com/QuarkNet-HEP/coding-camp/raw/refs/heads/main/data/dielectron_minus_bkgd.csv')
data.shape
```

```python
data.head(3)
```

```python
nbins = 100
plt.hist(data['mass'], bins=nbins, color='black', alpha=.5)
plt.show()
```

# Part 2: Clustering

The code below uses the k-means algorithm to identify clusters.

```python
#reformatting data for fitting
masses = data['mass'].to_numpy().reshape(-1,1)

gmm = GaussianMixture(n_components=2, covariance_type='spherical').fit(masses)
print('means:', gmm.means_)

x = np.linspace(0,100, num=1000)

n, bins, patches = plt.hist(masses, bins=100, alpha=.5)
for i in range(0, gmm.means_.shape[0]):
  plt.plot(x, 50000*stats.norm.pdf(x, gmm.means_[i], gmm.covariances_[i]**0.5))
  plt.vlines(gmm.means_[i],0,2000, colors='grey', linestyles='dotted')
plt.show()
```
