# Part 1: Plotting

The code below makes a  histogram of invariant mass for 14,000+ dielectron events with background subtracted.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv('https://github.com/QuarkNet-HEP/coding-camp/raw/refs/heads/main/data/dielectron_minus_bkgd.csv')
data.shape
```

```python
data.head(3)
```

```python
nbins = 100
plt.hist(data['mass'], bins=nbins, color='black', alpha=.5)
plt.show()
```

# Part 2: Clustering

The code below uses the k-means algorithm to identify clusters.

```python
kmeans = KMeans(n_clusters=2)
data['label'] = kmeans.fit_predict(data[['mass']])
print(kmeans.cluster_centers_)

for i in data['label'].unique():
  plt.hist(data[data['label']==i]['mass'], bins=nbins, alpha=.5, label=i)
plt.legend()
plt.show()
```
