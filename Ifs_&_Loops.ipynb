## If, Elif, and Else Blocks

A conditional statement allows the program to run a particular block of code depending on the input


If the condition is True, the code after the if statement runs. If it's false, the code after the else statement runs.

Here's a list of different [conditional tests](https://www.geeksforgeeks.org/relational-operators-in-python/) we can do.


Let's do a simple example below.

````python
# Let's try to create a choice in our code.

# create a variable that we'll use to determine the path
number = 5

# Let's start with a choice, if this is True, run the indented code below.
if number > 10:
  print(f'{number} is greater than 10')

# If the number wasn't greater than 10, run this indented code instead.
else:
  print(f'{number} is less than 10')
````

Change the number variable above and see how the output changes. Is everything working the way you expected?

### Adding in multiple checks

With our code above, there was no message for if the number was equal to ten. Here's a couple of ways we could fix that:
1. Change the if statement to be greater than or equal and modify the print statement
>```if number >= 10:```
2. Change the else block's print statement to be less than or equal to 10.
3. Or add an option to check if the number is equal to ten. Let's try that in the code block below.


````python
# Let's try to create a choice in our code.

# create a variable that we'll use to determine the path
number = 5

# Let's start with a choice, if this is True, run the indented code below.
if number > 10:
  print(f'{number} is greater than 10')

# If the number wasn't greater than 10, check if this is true.
# here we use == to see if the values are equal.
elif number == 10:
  print(f'{number} is equal to 10')

# And finally if both checks are False, run this code block.
else:
  print(f'{number} is less than 10')
````

## Loops

A loop allows us to run a block of code several times. This can be very useful when we want to generate data or do an operation on many different terms.




There are 2 main kinds of loops:
* A "For" Loop which runs the block of code a set number of times.
* A "While" Loop which runs the block of code until a condition is met.

### For Loops

````python
# Simple For Loop
# We use a variable (i) to keep track of the number of times the loop has happened
# The range(n) function generates a list of all the numbers from 0 to n-1

for i in range(10):  # This line starts the for loop.
  print(i) # The indented code will be run each time
````

Notice that we didn't actually print 10, but there were 10 different numbers printed. That's because Python (like most computer languages) starts counting from 0, not from 1.

Try to edit the code above to have the loop run 20 times.

````python
# Here's another for loop. Take a moment and see if you can figure out what it will print. Then run the code and find out.
a = 5
for i in range(10):
  print(a + i)
````

### While Loops

````python
# Let's recreate the first loop from the for section as a while loop

i = 0              # We need to define this variable before we start the loop
while (i < 10):    # Be sure to include the condition inside the (). The loop will stop when the condition is false.
  print(i)         # Indent the block of code we want to run each time.
  i += 1           # Increase the value of i by 1. If we don't do this, the loop will run forever.
````

````python
# Here's the second loop as well.
a = 5
i = 0
while (i < 10):
  print(a + i)
  i += 1
````

Try to change the code above to have the loop run 20 times. Notice the similarities between the for loop and the while loop.

### Which one do I choose?

Usually you should use a for loop if:
* You know exactly how many times the loop should run.
* You're going to loop over the items in a list. (see the code block below for an example)

````python
# Looping over a list.
# Let's say we want to add 10 to each number in the list.
my_list = [1,2,3,5,7]
for item in my_list:
  print(item + 10)
````

Usually you'll want to use a while loop if:
* You don't know how many times the loop should run
* You want the loop to run forever
