When you're ready, run each code cell until you get down to **Part Two**.

```python
# makes the scatterplot
plotdata = data.query('depth > 0')  # depth is in km

fig = px.scatter_geo(
    plotdata,
    lat='latitude',
    lon='longitude',
    projection='natural earth',
    opacity=0.6,
    title='Title!',              # try changing this
)
fig.update_traces(marker=dict(size=4, color='black'))
fig.update_geos(showcountries=True, showcoastlines=True, showland=True, landcolor='rgb(235,235,235)')
fig.update_layout(height=600)
fig.show()
```
